````python
from waves.solver import simulate
from waves.visualization import animate_solution
import matplotlib.pyplot as plt
````

## Conservative wave equation

`nu = gamma = 0` recovers the non-dissipative case: discrete energy should
stay constant to machine precision (printed below).

````python
result = simulate(N=201, c=1.0, T=1.5, cfl=0.4, store_every=4, nu=0.0, gamma=0.0)
````

## Viscous wave equation

Add Kelvin-Voigt damping (`nu`) and/or linear drag (`gamma`). Energy should
now decay monotonically -- this is checked automatically in `tests/test_energy.py`,
but it's nice to see and plot interactively too.

````python
result = simulate(
    N=201, c=1.0, T=10, cfl=0.4, store_every=4,
    nu=2e-4, gamma=0.0, theta=0.5,
)

plt.figure(figsize=(6, 4))
plt.plot(result.times, result.energies / result.energies[0])
plt.xlabel("t")
plt.ylabel(r"$E(t)/E(0)$")
plt.title("viscous energy decay")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## Animation

Renders an MP4 (falls back to GIF if FFmpeg isn't configured).

````python
animate_solution(
    result.X, result.Y, result.frames, result.times,
    filename="wave_carpet_viscous.mp4",
)
````
